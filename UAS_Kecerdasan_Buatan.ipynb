```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt

beban_sks = ctrl.Antecedent(np.arange(0, 25, 1), 'beban_sks')
tingkat_semester = ctrl.Antecedent(np.arange(1, 9, 1), 'tingkat_semester')

risiko_stres = ctrl.Consequent(np.arange(0, 101, 1), 'risiko_stres')

beban_sks['ringan'] = fuzz.trimf(beban_sks.universe, [0, 0, 15])
beban_sks['sedang'] = fuzz.trimf(beban_sks.universe, [12, 18, 22])
beban_sks['berat']  = fuzz.trimf(beban_sks.universe, [18, 24, 24])

tingkat_semester['rendah'] = fuzz.trimf(tingkat_semester.universe, [1, 1, 3])
tingkat_semester['sedang'] = fuzz.trimf(tingkat_semester.universe, [2, 4, 6])
tingkat_semester['tinggi'] = fuzz.trimf(tingkat_semester.universe, [5, 8, 8])

risiko_stres['rendah'] = fuzz.trimf(risiko_stres.universe, [0, 0, 40])
risiko_stres['sedang'] = fuzz.trimf(risiko_stres.universe, [30, 50, 70])
risiko_stres['tinggi'] = fuzz.trimf(risiko_stres.universe, [60, 100, 100])


rules = [
    ctrl.Rule(beban_sks['ringan'] & tingkat_semester['rendah'],
              risiko_stres['rendah']),

    ctrl.Rule(beban_sks['ringan'] & tingkat_semester['tinggi'],
              risiko_stres['sedang']),

    ctrl.Rule(beban_sks['sedang'] & tingkat_semester['sedang'],
              risiko_stres['sedang']),

    ctrl.Rule(beban_sks['sedang'] & tingkat_semester['tinggi'],
              risiko_stres['tinggi']),

    ctrl.Rule(beban_sks['berat'] & tingkat_semester['rendah'],
              risiko_stres['sedang']),

    ctrl.Rule(beban_sks['berat'] & tingkat_semester['sedang'],
              risiko_stres['tinggi']),

    ctrl.Rule(beban_sks['berat'] & tingkat_semester['tinggi'],
              risiko_stres['tinggi']),
]

stres_ctrl = ctrl.ControlSystem(rules)

semester = [1, 2, 3, 4, 5, 6, 7, 8]
sks = [22,20, 22, 20, 22, 20, 24, 6]

hasil_risiko = []

for i in range(len(semester)):
    simulasi = ctrl.ControlSystemSimulation(stres_ctrl)

    simulasi.input['beban_sks'] = sks[i]
    simulasi.input['tingkat_semester'] = semester[i]

    simulasi.compute()

    hasil = simulasi.output['risiko_stres']
    hasil_risiko.append(hasil)

    print(f"Semester {semester[i]} | SKS {sks[i]} | Risiko Stres: {hasil:.2f}")

plt.figure()
plt.plot(semester, hasil_risiko, marker='o')
plt.xlabel("Semester")
plt.ylabel("Risiko Stres")
plt.title("Risiko Stres Mahasiswa Berdasarkan Beban SKS")
plt.grid()
plt.show()
```
